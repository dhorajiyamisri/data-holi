```python
# 6. Handle Missing Data
import numpy as np

df.loc[0:20, "annual_income"] = np.nan
df.loc[30:40, "credit_score"] = np.nan
df.loc[50:60, "employment_type"] = np.nan

df.isnull().sum()
```

```python
# Simple Imputer (Numerical Mean, Median) 

from sklearn.impute import SimpleImputer

mean_imputer = SimpleImputer(strategy="mean")

df["annual_income"] = mean_imputer.fit_transform(
    df[["annual_income"]]
).ravel()


df["annual_income"].isnull().sum()
```

```python
# mode
median_imputer = SimpleImputer(strategy="median")

df["credit_score"] = median_imputer.fit_transform(
    df[["credit_score"]]
).ravel()


df["credit_score"].isnull().sum()
```

```python
# Categorical Most Frequent
cat_imputer = SimpleImputer(
    strategy="most_frequent"
)

df["employment_type"] = cat_imputer.fit_transform(
    df[["employment_type"]]
).ravel()

df["employment_type"].isnull().sum()
```

```python
# Most Frequent Category Imputation
df["gender"] = df["gender"].fillna(
    df["gender"].mode()[0]
)

print(df)
```

```python
# Missing Indicator
df.loc[0:10, "annual_income"] = np.nan

df["income_missing"] = (
    df["annual_income"]
    .isnull()
    .astype(int)
)
print(df[["annual_income","income_missing"]].head(15))

# Complete Case Analysis
df_complete = df.dropna()

print(df_complete.shape)

# Random Sample Imputation
random_sample = (
    df["annual_income"]
    .dropna()
    .sample(
        df["annual_income"].isnull().sum(),
        random_state=42
    )
)
random_sample.index = df[
    df["annual_income"].isnull()
].index
df.loc[
    df["annual_income"].isnull(),
    "annual_income"
] = random_sample

print(df["annual_income"].isnull().sum())

# KNN Imputer
df.loc[0:10, "loan_amount"] = np.nan
from sklearn.impute import KNNImputer
knn = KNNImputer(n_neighbors=5)
cols = [
    "annual_income",
    "loan_amount",
    "credit_score"
]
df[cols] = knn.fit_transform(df[cols])
print(df[cols].head())

# MICE Imputer
df.loc[15:25, "credit_score"] = np.nan

from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
mice = IterativeImputer(random_state=42)
df[cols] = mice.fit_transform(df[cols])

print(df[cols].head())
```

```python
                    # part D : outlier handling

# Z-Score Method

from scipy import stats
z = np.abs(stats.zscore(df["annual_income"]))
df_z = df[(z < 3)]
print(df_z)
```

```python
from scipy.stats.mstats import winsorize
import numpy as np

winsorized_data = winsorize(
    df["annual_income"],
    limits=[0.05, 0.05]
)

df["annual_income"] = np.array(winsorized_data)
print("Before Winsorization")
print(df["annual_income"].describe())

from scipy.stats.mstats import winsorize
import numpy as np

df["annual_income"] = np.array(
    winsorize(
        df["annual_income"],
        limits=[0.05,0.05]
    )
)

print("\nAfter Winsorization")
print(df["annual_income"].describe())
```

```python
lower = df['annual_income'].quantile(0.01)

upper = df['annual_income'].quantile(0.99)

print("1st Percentile :", lower)
print("99th Percentile :", upper)

df_percentile = df[
    (df['annual_income'] >= lower) &
    (df['annual_income'] <= upper)
]

print("\nOriginal Shape :", df.shape)
print("After Percentile Method :", df_percentile.shape)
```

```python
# Part E : Feature Engineering
# Handle Variable Types

print(df.dtypes)
```

```python
df["join_date"] = pd.to_datetime(df["join_date"])

# Year, Month, Day, Weekday extract karo:

df["Year"] = df["join_date"].dt.year
df["Month"] = df["join_date"].dt.month
df["Day"] = df["join_date"].dt.day
df["Weekday"] = df["join_date"].dt.day_name()

print(df[["join_date","Year","Month","Day","Weekday"]].head())
```

```python
# Ordinal Encoding
from sklearn.preprocessing import OrdinalEncoder

oe = OrdinalEncoder()

df["education_level_encoded"] = oe.fit_transform(
    df[["education_level"]]
)

print(
    df[
        ["education_level",
         "education_level_encoded"]
    ].head()
)
```

```python
# Label Encoding 
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df["gender_encoded"] = le.fit_transform(
    df["gender"]
)

print(
    df[
        ["gender",
         "gender_encoded"]
    ].head()
)
```

```python
# One Hot Encoding 
region_ohe = pd.get_dummies(
    df["region"],
    prefix="region"
)

print(region_ohe.head())
```

```python
# Binning
df["income_bin"] = pd.cut(
    df["annual_income"],
    bins=3,
    labels=["Low","Medium","High"]
)

print(
    df[
        ["annual_income",
         "income_bin"]
    ].head()
)
```

```python
# Binarization
from sklearn.preprocessing import Binarizer

binarizer = Binarizer(
    threshold=500000
)

df["income_binary"] = binarizer.fit_transform(
    df[["annual_income"]]
)

print(
    df[
        ["annual_income",
         "income_binary"]
    ].head()
)
```

```python
# quantile Binning
df["income_quantile"] = pd.qcut(
    df["annual_income"],
    q=4,
    labels=False
)

print(
    df[
        ["annual_income",
         "income_quantile"]
    ].head()
)
```

```python
# K-Means Binning
from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=3,
    random_state=42
)

df["income_cluster"] = kmeans.fit_predict(
    df[["annual_income"]]
)

print(
    df[
        ["annual_income",
         "income_cluster"]
    ].head()
)
```

```python
# Part F : Feature Scaling
# Standardization
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df["income_standard"] = scaler.fit_transform(
    df[["annual_income"]]
)

print(
    df[
        ["annual_income",
         "income_standard"]
    ].head()
)
```

```python
# Normalization
from sklearn.preprocessing import Normalizer

normalizer = Normalizer()

df[["annual_income","loan_amount"]] = normalizer.fit_transform(
    df[["annual_income","loan_amount"]]
)

print(
    df[
        ["annual_income",
         "loan_amount"]
    ].head()
)
```

```python
# Min-Max Scaling
from sklearn.preprocessing import MinMaxScaler

mms = MinMaxScaler()

df["income_minmax"] = mms.fit_transform(
    df[["annual_income"]]
)

print(
    df[
        ["annual_income",
         "income_minmax"]
    ].head()
)
```

```python
# MaxAbs Scaling
from sklearn.preprocessing import MaxAbsScaler

maxabs = MaxAbsScaler()

df["income_maxabs"] = maxabs.fit_transform(
    df[["annual_income"]]
)

print(
    df[
        ["annual_income",
         "income_maxabs"]
    ].head()
)
```

```python
# Robust Scaling
from sklearn.preprocessing import RobustScaler

robust = RobustScaler()

df["income_robust"] = robust.fit_transform(
    df[["annual_income"]]
)

print(
    df[
        ["annual_income",
         "income_robust"]
    ].head()
)
```
